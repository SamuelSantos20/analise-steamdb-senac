```python
import time
import polars as pl

# 1. Leitura Lazy: prepara a consulta sem carregar todos os dados na RAM
q = (
    pl.scan_parquet("steam_2026_organizada_PANDAS.parquet")
    .filter(pl.col("Lançamento").dt.year() == 2025)
)

print(type(q))  # LazyFrame: a consulta ainda não foi executada
print("\n")

# 2. Plano de execução da consulta
print("--- Plano de execução: jogos lançados em 2025 ---")
print("Veja como o Polars planeja executar e otimizar a consulta:")
print(q.explain())
print("\n")

# 3. Executa a consulta e carrega o resultado
print("--- Consultando os jogos da Steam ---")
start_time = time.perf_counter()
df_resultado = q.collect()
end_time = time.perf_counter()

print(f"Quantidade de jogos lançados em 2025: {df_resultado.height}")
print(f"Tempo de execução: {end_time - start_time:.2f} segundos")

display(df_resultado)
```
