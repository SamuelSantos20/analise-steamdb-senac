```python
import polars as pl
import matplotlib.pyplot as plt


df = pl.read_parquet('steam_2026_organizada_PANDAS.parquet')

resultado = (
    df.with_columns(
        pl.col('Lançamento')
        .dt.year()
        .alias('ano')
    )
    .group_by('ano')
    .len()
    .sort('ano')
)

anos = resultado['ano'].to_list()
quantidades = resultado['len'].to_list()

plt.bar(anos, quantidades)

plt.xlabel('ANO DE LANÇAMENTO')
plt.ylabel('QUANTIDADE DE LANÇAMENTOS')
plt.title('N. DE LANÇAMENTOS POR ANO')

plt.show()
```
